```python
# Kütüphane

# scikit-learn toy datasets modülünden hazır Iris veri setini getiren fonksiyon
from sklearn.datasets import load_iris

#Veriyi train ve test'e bölen fonksiyon
from sklearn.model_selection import train_test_split

#Tek bir karar ağacı kuran sınıflandırma modeli
from sklearn.tree import DecisionTreeClassifier

#Birden fazla karar ağacını birlikte çalıştırır ensemble modeli
from sklearn.ensemble import RandomForestClassifier

#Sınıflandırma tahminlerini değerlendiren temel metrik fonksiyonları
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

1 Iris Dataset Yükleme ve Features Target Ayrımı

```python

# UCI Repository'de yer alan 150 örnek ve 3 çiçek türünden oluşan klasik Iris
#veri setini scikit-learn toy datasets içindeki load_iris() ile yüklendi
iris_verisi = load_iris()

#Features (özellikler-girdi değişkeni)
# Modelin karar verirken kullanacağı sayısal sütunlardır
# yaprak uzunluk ve genişlik ölçümleri
girdi_ozellikleri = iris_verisi.data

#Target (hedef-çıktı değişkenleri)
#Modelin öğrenip tahmin yapmaya çalışacağı sınıf etiketidir 0,1,2 gibi
hedef_siniflar = iris_verisi.target

#Veri yapısını doğrulama (boyut ve isim yazdırma)
print("1. Veri Seti Özeti")
print("Features isimleri: ", iris_verisi.feature_names)
print("Target isimleri: ", list(iris_verisi.target_names))
print( "Özellik matrisi boyutu: ", girdi_ozellikleri.shape)
print("Hedef vektörü boyutu: ", hedef_siniflar.shape)


#ÇIKTI
# 4 özellik sütununun (sepal/petal uzunluk ve genişlikleri) yüklendiğini gösterir
# Hedef sınıfların 3 çiçek türü (setosa, versicolor, virginica) olduğunu doğrular
#(150, 4) ve (150,) boyutları, toplam 150 çiçek örneğinin 4 sayısal özellik ve
#1 hedef etiketle eksiksiz geldiğini gösterir
```

2 train_test_split

```python
# Kagle üzerinden anlatılan konuda model eğitim verisini ezberleyebileceği
#için veriyi iki parçaya böleriz;
#girdi_egitim, hedef_egitim modelin kuralları öğrenebileceği %75 eğitim seti
#girdi_test, hedef_test modelin daha önce görmediği %25lik test seti
#random_state = 7 bölme işleminin her adımda aynı çıkmasını
#stratify=hedef_siniflar her türün iki tarafta da eşit bulunmasını
#sağlar

girdi_egitim, girdi_test, hedef_egitim, hedef_test = train_test_split(
    girdi_ozellikleri, hedef_siniflar, test_size=0.25,
    random_state=7,stratify=hedef_siniflar
)

print("Train Test Ayrımı")
print("Eğitim Seti Örnek Sayısı:", len(girdi_egitim))
print("Test Seti Örnek Sayısı:", len(girdi_test))

# ÇIKTI
# 150 örneğin %75'ine karşılık gelen 112 örneğin modeli eğitmek için ayrıldığını,
#%25'ine karşılık gelen 38 örneğin ise modelin hiç görmediği test verisi olarak
#ayrıldığını ve genelleme başarımının bu 38 örnek üzerinden ölçüleceğini anlatır
```

3 Decision Tree Modelini Kurma ve Prediction(Tahmin)

```python
# Decision Tree(Karar Ağacı):Özellik değerlerine göre dallanan karar
#kuralları oluşturarak sınıflandırma yapan model

agac_modeli = DecisionTreeClassifier(random_state=7)

#.fit ile modeli sadece eğitim verisi eğitilir
agac_modeli.fit(girdi_egitim, hedef_egitim)

#Prediction eğitilmiş karar ağacının .predict ile test verisindeki
#(girdi_test)çiçek ölçümlerine bakarak tür tahmini üretmesidir
agac_tahmini = agac_modeli.predict(girdi_test)
```

4 Random Forest Modelini Kurma ve Prediction

```python
# Random forest tek karar ağacına güvenmek yerine biirden fazla karar ağacını
#(n_estimatores=50 ile adet) birlikte eğitir ve tahminleri çoğunluğa göre
#birleştirerek daha güçlü bir ağaç modeli kurar
orman_modeli = RandomForestClassifier(n_estimators=50, random_state=7)

# Burada .fit oramandaki 50 ağacı aynı eğitim veri setiyle eğitir
orman_modeli.fit(girdi_egitim, hedef_egitim)

# Prediction eğitilmiş random forest modelini aynı test verisi yani girdi_test
#üzerinden sınıf tahmini üretir
orman_tahmini = orman_modeli.predict(girdi_test)


print("Predictionlar")
print("Gerçek Sınıflar: ", list(hedef_test[:8]))
print("Decision Tree Tahmin: ", list(agac_tahmini[:8]))
print("Random Forest Tahmini: ", list(orman_tahmini[:8]))


# ÇIKTI
# Test setindeki ilk 8 örneğin gerçek sınıf etiketleriyle (0, 0, 2, 1, 0, 1, 0, 2)
#hem Decision Tree hem de Random Forest tahminlerinin birebir örtüştüğünü gösterir
# İki modelin de bu ilk 8 test örneğinde hatasız tahmin ürettiğini gösterir
```

5 Temel Metriklerle Karşılaştırma

```python
#Modelin test tahminlerini tek tek karşılaştırmak için sözlük kurma
karsılastırma_listesi = {
    "Decision Tree":agac_tahmini,
    "Random Forest":orman_tahmini
}
print("Temel Sınıflandırma Metrikleri Karşılaştırması")
for model_adi, model_tahmini in karsılastırma_listesi.items():

# Accuary -doğruluk- test setindeki tüm tahminler içinde doğru bilinen
# örneklerin toplam örnek sayısına oranıdır.
    dogruluk = accuracy_score(hedef_test, model_tahmini)

# Precision kesinlik modelin belirli bir sınıfa ait diye tahmin ettiği
#örneklerin gerçekte ne kadarının o sınıfa ait olduğunu ölçer.
#(average="weighted": 3 çiçek sınıfının ağırlıklı ortalamasını hesaplar)
    kesinlik = precision_score(hedef_test, model_tahmini, average = "weighted")


# Recall duyarlılık gerçekte o sınıfta bulunan örneklerin ne
#kadarının model tarafından doğru yakalandığını ölçer
    duyarlilik = recall_score(hedef_test, model_tahmini, average="weighted")

#F1 Score Precision ve Recall metriklerini dengeli şekilde tek bir
#skorda birleştiren ortalama ölçütü
    f1_skoru = f1_score(hedef_test, model_tahmini, average="weighted")

    print(f"\n[{model_adi}]")
    print(f"Accuracy: {dogruluk:.4f}")
    print(f"Precision: {kesinlik:.4f}")
    print(f"Recall: {duyarlilik:.4f}")
    print(f"F1 Score: {f1_skoru:.4f}")


# ÇIKTI
# Decision Tree modeli test setinde 0.9474 Accuracy ve 0.9472 F1 Score ile
#çok yüksek bir başarı elde etmiş
# Random Forest modeli ise 0.9737 Accuracy ve 0.9737 F1 Score ile tek bir
#karar ağacına kıyasla test verisinde daha yüksek genelleme başarımı göstermiştir
```

   Iris çiçeğinin türünü (`setosa`, `versicolor`, `virginica` — sayısal olarak `0`, `1`, `2` hedef sınıflarını) tahmin ediyor.

2. Hangi veriden öğreniyor?

   Iris veri setindeki 150 örneğin eğitim setine ayrılan 112 örneğinden (%75 train seti) ve çiçeklerin çanak/taç yaprak uzunluk-genişlik ölçümlerini içeren 4 sayısal özellik sütunundan (sepal length (cm), sepal width (cm), petal length (cm), petal width (cm)) ve bu örneklere karşılık gelen gerçek sınıf etiketlerinden öğreniyor.

3. Tahminini neye göre başarılı kabul ettin?

   Modelin eğitimde görmediği 38 örnekten oluşan test seti (%25 test seti) üzerinde ürettiği tahminlerin gerçek etiketlerle örtüşme düzeyini Accuracy, Precision, Recall ve F1 Score metrikleriyle ölçerek başarılı kabul ettim. Elde edilen sonuçlara göre Decision Tree modeli `0.9474` Accuracy ve `0.9472` F1 Score üretirken, Random Forest modeli `0.9737` Accuracy ve `0.9737` F1 Score ile daha yüksek bir genelleme başarımı gösterdi.
