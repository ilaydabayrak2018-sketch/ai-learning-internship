1- Sayı Ortalaması Hesaplama

```python
# for döngüsü ile liste elemanları üzerinde tek tek dolaşabiliriz.
#  makine öğrenmesinde kayıp (loss) değerlerinin veya veri kümesindeki sayısal sütunların ortalamasını alırken bu döngü ve toplam mantığının temel oluşturur
# Boş veri gelme ihtimaline karşı koşul yapısı (if/else) ile kontrol mekanizması kurmanın kodun sağlamlığı açısından önemli.

# sayı listesi
sayilar = [12, 45, 68, 23, 90, 54, 71]
toplam = 0
adet = 0

# değişkenleri tanımlama

# listenin boş olup olmadığını kontrol ediyoruz
if len(sayilar) > 0:
    # her sayıyı gezmek için for kullanıyoruz
    for sayi in sayilar:
        toplam += sayi
        adet += 1

    # ortalama hesaplama
    ortalama = toplam / adet
    print(f"Sayıların Toplamı: {toplam}")
    print(f"Eleman Sayısı    : {adet}")
    print(f"Aritmetik Ortalama: {ortalama:.2f}")
else:
    # sıfıra bölme hatasını önler
    print("Hata: Ortalama hesaplanacak listede eleman bulunmuyor!")

# Bu programda for döngüsü ile veri akışını toplamayı ve listenin boş olma riskini if kontrolüyle yöneterek hata yönetimini sağladım.
```

2- Not Sınıflandırma

```python
# notları 0-100 aralığına göre harf notuna dönüştürür ve harf notuna göre öğrenciye özel geri bildirim mesajı üretir.
# if, elif ve else sıralaması
# geçersiz veri girişlerini örneğin negatif veya 100'den büyük notlar için filtreleme

def not_degerlendir(puan):
    #giriş verisinin geçerlilik aralığını kontrol ediyor
    if puan < 0 or puan > 100:
        return "Geçersiz Puan"

    #if-elif-else ile harf notunu buluyor
    if puan >= 90:
        harf = "A"
    elif puan >= 80:
        harf = "B"
    elif puan >= 70:
        harf = "C"
    elif puan >= 60:
        harf = "D"
    else:
        harf = "F"

    return harf

# deneme için puan tanımlama
ogrenci_puani = 85
harf_sonucu = not_degerlendir(ogrenci_puani)

print(f"Öğrenci Puanı: {ogrenci_puani} -> Harf Notu: {harf_sonucu}")

# match-case ile elde edilen harf sonucuna göre değerlendirme
match harf_sonucu:
    case "A":
        print("Sonuç: Üstün Başarı! Dersi mükemmel dereceyle tamamladınız.")
    case "B" | "C":
        print("Sonuç: Tebrikler, dersi başarılı bir şekilde geçtiniz.")
    case "D":
        print("Sonuç: Şartlı / Sınırda Geçti. Konuları tekrar etmeniz önerilir.")
    case "F":
        print("Sonuç: Kaldı. Dersi tekrar almanız gerekmektedir.")
    case _:
        print("Sonuç: Değerlendirme yapılamadı.")

# Çoklu karar mekanizmalarında if-elif bloklarıyla sürekli aralıkları sınıflandırmayı,match-case yapısıyla da ayrık kategorileri temiz ve okunabilir şekilde yönetmeyi öğrendim.
```

4-calculate_average(numbers) Fonksiyonu ve Hata Denetimi

```python
# parametre olarak aldığı sayı listesinin ortalamasını hesaplayıp çağıran noktaya döndürür (return).
#sayısal olmayan tipleri, boş listeleri ve hatalı parametreleri try-except ile filtreler.

# fonksiyonda 'def', ve 'return' kod tekrarını engelleyen en temel modüler yapıdır.
# fonksiyonun sadece beklenen durumlarda değil liste içinde string gibi uyumsuz tipler geldiğinde (TypeError) veya boş liste geldiğinde nasıl davranması gerektiğini
#try-except bloklarıyla kontrol altına alır.

def calculate_average(numbers):
    """
    Verilen sayı listesinin aritmetik ortalamasını güvenli bir şekilde hesaplar.
    Hatalı veri tiplerinde veya boş dizilerde programın çökmesini engeller.
    """
    try:
        # listenin tip kontrolü
        if not isinstance(numbers, list):
            raise TypeError("parametre mutlaka bir liste (list) olmalıdır!")

        # boş liste kontrolü
        if len(numbers) == 0:
            print("gönderilen liste boş, ortalama 0 olarak döndürülüyor.")
            return 0.0

        # elemanların toplamı
        toplam = sum(numbers)
        ortalama = toplam / len(numbers)
        return ortalama

    except TypeError as hata:
        print(f"Hata Tespit Edildi: {hata}")
        return None
    except Exception as beklenmeyen_hata:
        print(f"Beklenmeyen bir sorun oluştu: {beklenmeyen_hata}")
        return None

# farklı senaryo testi

print("Senaryo 1 (Normal Liste):")
notlar = [85, 90, 78, 92, 100]
sonuc1 = calculate_average(notlar)
print(f"Sonuç: {sonuc1}\n")

print("Senaryo 2 (Boş Liste Testi):")
bos_liste = []
sonuc2 = calculate_average(bos_liste)
print(f"Sonuç: {sonuc2}\n")

print("Senaryo 3 (Uyumsuz Veri Tipi Testi - Hata Yönetimi):")
hatali_liste = [10, 20, "otuz", 40]
sonuc3 = calculate_average(hatali_liste)
print(f"Sonuç: {sonuc3}\n")


# calculate_average fonksiyonu ile fonksiyonların yeniden kullanılabilirliğini (reusability)
# ve try-except bloklarıyla beklenmedik veri girişlerine karşı fonksiyonun nasıl korunduğunu test ettim.
```

5- While Döngüsü

```python
#while döngüsü belirli bir koşul sağlandığı sürece çalışmaya devam eder
#veri modelinde en üst .atı olan data science ile başlayan en alt dala kadar gezinmek için idealdir
#döngüyü burada sonsuza girmemesi için her adımda işaretçiyi alt seviyeye güncelleriz

ai_hiyerarsi = {
    "kavram" : "Data Science" ,
    "alt_kavram": {
        "kavram" : "AI" ,
        "alt_kavram": {
            "kavram" : "Machine learning" ,
            "alt_kavram":{
                "kavram" : "Deep Learning" ,
                "alt_kavram":{
                    "kavram" : "Generative AI" ,
                    "alt_kavram" : {
                        "kavram" : "Large Language Models",
                        "alt_kavram" :None
                    }
                }
            }
        }
    }
}

adim = 1
gecici_dugum = ai_hiyerarsi  #bu satırla döngü başlamadan önce en üstteki sözlüğe (Data Science) konumlanır

print("While Döngüsü ile Kavram Haritası")

while gecici_dugum is not None:
  print(f"{adim}. Seviye: {gecici_dugum ['kavram']}")
  #bir alt basamağa geçiş
  gecici_dugum = gecici_dugum["alt_kavram"]
  adim += 1

# Bu çalışmada while döngüsünü sayaç bazlı gezinmenin yanı sıra
# koşula bağlı veri yapılarında derinlemesine ilerlemek
# amacıyla kullandım.
```

7- Dosya Okuma ve Yazma

```python
# 'with open' bağlam yöneticisi , dosya açma-kapama işlemlerini
# güvenli hale getirir ve olası bellek sızıntılarını önler.
# 'w' moduyla kavram hiyerarşisi diske yazılır ve
# 'r' moduyla doğrulanmak üzere geri okunur.

dosya_adi = "kavram_harita.txt" #boş bir dosya oluştu

harita_adımları = [
    "1-Data Science \n",
    "2-AI\n",
    "3-Machine Learning\n",
    "4-Deep Learning\n",
    "5-Computer Vision & NLP\n",
    "6-Generative AI\n",
    "7-Large Language Model\n"
]

#dosya yazma - writing
#encoding="utf-8" Türkçe karakterlerin bozulmadan doğru kaydedilmesini ve okunmasını sağlar
# 'w' moduyla dosya açılır (yoksa oluşturulur, varsa sıfırlanır) ve liste satırları yazılır
with open(dosya_adi, "w" , encoding = "utf-8") as dosya:

    dosya.writelines(harita_adımları) # Listedeki tüm elemanları topluca dosyaya aktarır
print(f"' {dosya_adi}' dosya kaydedildi. \n")

print("Dosyadan Okunan Kavramlar")
#'r' moduyla dosya salt okunur olarak açılır ve satır satır taranır
with open(dosya_adi, "r" , encoding="utf-8") as dosya:
  for satir in dosya:
    print(satir.strip()) #satır sonundaki görünmez '\n' boşluklarını temizleyerek ekrana basar


#with open yapısıyla dosya okuma ve yazma işlemlerinde
#kaynak yönetimini, dosya modlarını ('w', 'r')
#ve karakter kodlaması (utf-8) standartlarını uyguladım.
```

8- Dictionary .get() Metodu ile Güvenli Veri Sorgulama

```python
# Sözlüklerde doğrudan köşeli parantez my_dict['anahtar'] kullanır
#anahtar mevcut değilse KeyError hatası fırlatır.
# .get() metodu ise anahtar bulunamadığında hata vermek yerine
#None veya tanımlanan varsayılan değeri güvenli bir şekilde döndürür.

kavram_harita_aciklama ={
    "Data Science": "Veriden anlam çıkarma ve analiz etme disiplini.",
    "Artificial Intelligence": "Akıllı karar verebilen sistemler bütünü.",
    "Machine Learning": "Veriye dayalı örüntü öğrenme yöntemleri.",
    "Deep Learning": "Yapay sinir ağları odaklı makine öğrenimi dalı."
}

sorgulanacak_kavramlar =[
    "Data Science",
    "Computer Vision",
    "Deep Learning",
    "Large Language Models"
]

print(".get()ile Sözlük Sorgulama ")
for kavram in sorgulanacak_kavramlar:
# get(anahtar, varsayilan_deger) kullanımı

    aciklama = kavram_harita_aciklama.get(kavram, "Bu kavramın açıklaması bulunmamaktadır.")
    print(f"Kavram: {kavram}")
    print(f"Açıklama: {aciklama} \n")


# .get() metodunu kullanarak sözlük sorgularında istisna yönetimini sağlamayı ve
# KeyError kaynaklı program kesintilerini önlemeyi çalıştım.
```
