1- Sayı Ortalaması Hesaplama

```python
# for döngüsü ile liste elemanları üzerinde tek tek dolaşabiliriz.
#  makine öğrenmesinde kayıp (loss) değerlerinin veya veri kümesindeki sayısal sütunların ortalamasını alırken bu döngü ve toplam mantığının temel oluşturur
# Boş veri gelme ihtimaline karşı koşul yapısı (if/else) ile kontrol mekanizması kurmanın kodun sağlamlığı açısından önemli.

# sayı listesi
sayilar = [12, 45, 68, 23, 90, 54, 71]

# değişkenleri tanımlama
toplam = 0
adet = 0

# listenin boş olup olmadığını kontrol ediyoruz
if len(sayilar) > 0:
    # her sayıyı gezmek için for kullanıyoruz
    for sayi in sayilar:
        toplam += sayi
        adet += 1

    # ortalama hesaplama
    ortalama = toplam / adet
    print(f"Sayıların Toplamı: {toplam}")
    print(f"Eleman Sayısı    : {adet}")
    print(f"Aritmetik Ortalama: {ortalama:.2f}")
else:
    # sıfıra bölme hatasını önler
    print("Hata: Ortalama hesaplanacak listede eleman bulunmuyor!")

# Bu programda for döngüsü ile veri akışını toplamayı ve listenin boş olma riskini if kontrolüyle yöneterek hata yönetimini sağladım.
```

2- Not Sınıflandırma

```python
# notları 0-100 aralığına göre harf notuna dönüştürür ve harf notuna göre öğrenciye özel geri bildirim mesajı üretir.
# if, elif ve else sıralaması
# geçersiz veri girişlerini örneğin negatif veya 100'den büyük notlar için filtreleme

def not_degerlendir(puan):
    #giriş verisinin geçerlilik aralığını kontrol ediyor
    if puan < 0 or puan > 100:
        return "Geçersiz Puan"

    #if-elif-else ile harf notunu buluyor
    if puan >= 90:
        harf = "A"
    elif puan >= 80:
        harf = "B"
    elif puan >= 70:
        harf = "C"
    elif puan >= 60:
        harf = "D"
    else:
        harf = "F"

    return harf

# deneme için puan tanımlama
ogrenci_puani = 85
harf_sonucu = not_degerlendir(ogrenci_puani)

print(f"Öğrenci Puanı: {ogrenci_puani} -> Harf Notu: {harf_sonucu}")

# match-case ile elde edilen harf sonucuna göre değerlendirme
match harf_sonucu:
    case "A":
        print("Sonuç: Üstün Başarı! Dersi mükemmel dereceyle tamamladınız.")
    case "B" | "C":
        print("Sonuç: Tebrikler, dersi başarılı bir şekilde geçtiniz.")
    case "D":
        print("Sonuç: Şartlı / Sınırda Geçti. Konuları tekrar etmeniz önerilir.")
    case "F":
        print("Sonuç: Kaldı. Dersi tekrar almanız gerekmektedir.")
    case _:
        print("Sonuç: Değerlendirme yapılamadı.")

# Çoklu karar mekanizmalarında if-elif bloklarıyla sürekli aralıkları sınıflandırmayı,match-case yapısıyla da ayrık kategorileri temiz ve okunabilir şekilde yönetmeyi öğrendim.
```

```python
#   Key-value (anahtar-değer) kullanarak yapılandırılmış müşteri verilerini tutar.
#   yeni veri ekleme, veri güncelleme ve hata almadan veri sorgulama işlemlerini yürütür.

# müşteri kaydı bilgileri
musteri_kaydi = {
    "musteri_id": "M123",
    "ad": "İlayda",
    "soyad": "Bayrak",
    "departman": "Yönetim Bilişim Sistemleri",
    "aktif_mi": True,
    "harcamalar": [150.0, 420.5, 89.0]
}

print("--- MEVCUT MÜŞTERİ BİLGİLERİ ---")
#sözlükteki anahtar ve değerleri for döngüsüyle gezme (.items() kullanımı)
for anahtar, deger in musteri_kaydi.items():
    print(f"{anahtar.capitalize()}: {deger}")

# yeni bilgi ekleme
musteri_kaydi["sehir"] = "İstanbul"
musteri_kaydi["harcamalar"].append(310.0)  # Listenin içine yeni harcama ekleme


# müşterinin toplam harcamasını hesaplar
toplam_harcama = sum(musteri_kaydi["harcamalar"])
print(f"Toplam Harcama Tutarı: {toplam_harcama} TL")


#dictionary yapısı üzerinden karmaşık verilerin nasıl organize edildiğini, veri okurken çökmeleri önlemek için .get() kullanımının kritik önemini kavradım.
```

4-calculate_average(numbers) Fonksiyonu ve Hata Denetimi

```python
# parametre olarak aldığı sayı listesinin ortalamasını hesaplayıp çağıran noktaya döndürür (return).
#sayısal olmayan tipleri, boş listeleri ve hatalı parametreleri try-except ile filtreler.

# fonksiyonda 'def', ve 'return' kod tekrarını engelleyen en temel modüler yapıdır.
# fonksiyonun sadece beklenen durumlarda değil liste içinde string gibi uyumsuz tipler geldiğinde (TypeError) veya boş liste geldiğinde nasıl davranması gerektiğini
#try-except bloklarıyla kontrol altına alır.

def calculate_average(numbers):
    """
    Verilen sayı listesinin aritmetik ortalamasını güvenli bir şekilde hesaplar.
    Hatalı veri tiplerinde veya boş dizilerde programın çökmesini engeller.
    """
    try:
        # listenin tip kontrolü
        if not isinstance(numbers, list):
            raise TypeError("parametre mutlaka bir liste (list) olmalıdır!")

        # boş liste kontrolü
        if len(numbers) == 0:
            print("gönderilen liste boş, ortalama 0 olarak döndürülüyor.")
            return 0.0

        # elemanların toplamı
        toplam = sum(numbers)
        ortalama = toplam / len(numbers)
        return ortalama

    except TypeError as hata:
        print(f"Hata Tespit Edildi: {hata}")
        return None
    except Exception as beklenmeyen_hata:
        print(f"Beklenmeyen bir sorun oluştu: {beklenmeyen_hata}")
        return None

# farklı senaryo testi

print("Senaryo 1 (Normal Liste):")
notlar = [85, 90, 78, 92, 100]
sonuc1 = calculate_average(notlar)
print(f"Sonuç: {sonuc1}\n")

print("Senaryo 2 (Boş Liste Testi):")
bos_liste = []
sonuc2 = calculate_average(bos_liste)
print(f"Sonuç: {sonuc2}\n")

print("Senaryo 3 (Uyumsuz Veri Tipi Testi - Hata Yönetimi):")
hatali_liste = [10, 20, "otuz", 40]
sonuc3 = calculate_average(hatali_liste)
print(f"Sonuç: {sonuc3}\n")


# calculate_average fonksiyonu ile fonksiyonların yeniden kullanılabilirliğini (reusability)
# ve try-except bloklarıyla beklenmedik veri girişlerine karşı fonksiyonun nasıl korunduğunu test ettim.
```
