1-VERİYİ YÜKLEME VE BOYUT KONTROLü

```python
import pandas as pd

#colab'e yüklediğim titanic verisinin csv dosyası okunuyor
#Passenger Id sütununu satır indeksi yapıyoruz (index_col ='PassengeId)

dosya_adi = "train_and_test2.csv"
df = pd.read_csv(dosya_adi, index_col="Passengerid")

#Burada csv dosytasını pandas dataframe'e aktardım
#satırlar yolcuları,sütunlar is yolculara ait özellikleri tutuyor

#satır sütun sayısını verir
print("Veri seti boyutu (Satır,Sütun):" , df.shape)

#ilk 5 satırı inceler
print("İlk 5 Gözlem:")
display(df.head())

```

2-VERİ TİPLERİ (DATA TYPES) VE EKSİK VERİ (MISSING VALUES) KONTROLÜ

```python
#her sütunun veri tipini topluca kontrol eder
print("Sütun Veri Tipleri")
print(df.dtypes)

#her sütundaki eksik değer saysını hesaplar
print("\n Eksik Veri Sayısı")
eksik_sayisi=df.isnull().sum()
print(eksik_sayisi[eksik_sayisi>0])
#burada boşluklar bulunuyor. Boşluklar modelde kullanılamaz
#ya doldurulmalı ya da satır sütun bazında eleme yapılmalı
```

3-İSTATİSTİKSEL ÖZETLER VE FREKANSLAR (DESCRIBE & VALUE_COUNTS)

```python
#sayısal sütunların özet istatistiklerini hesaplar
print("\n Sayısal Değişken İstatistiksel Özeti ")
display(df.describe())
#çıktı yorumlama
#burada yaş ortalaması 29.503186
#Fare min 0, max 512.329200	çıkmış bu değer aykırı değer olabileceğini gösteriyor


# Kategorik değişkenlerdeki sınıfların sayımı
print("\n Cinsiyet Dağılımı")
print(df['Sex'].value_counts())


#tekrar etmeyenlerin listesi
print("\n Benzersizler")
print(df['Embarked'].unique())
```

4-SATIR/SÜTUN İNDEKSLERİ VE FİLTRELEME (LOC, ILOC, BOOLEAN INDEXING)

```python
# iloc ile ilk 5 satır ve ilk 3 sütunu konuma göre seçer
print("\n iloc ile ilk 5 satır ve 3 sütun")
display(df.iloc[:5, :3])

# loc ve Boolean Indexing ile çoklu şartlı filtreleme
#1.sınıfta olan ve hayatta kalan yolcuları listeler
kurtulan_birinci_sinif = df.loc[(df['Pclass'] == 1) & (df['2urvived'] ==1)]
print(f"\n1. Sınıf Hayatta Kalan Sayısı: {len(kurtulan_birinci_sinif)}")


# isin() kullanarak filtreleme
liman_cq = df.loc[df['Embarked'].isin([1, 2]), ['Age', 'Fare']]
print("\n C ve Q Limanından Binen İlk 3 Yolcu")
display(liman_cq.head(3))
```

5-VERİ TEMİZLEME VE DÖNÜŞTÜRME (FILLNA, REPLACE, ASTYPE)

```python
# Orijinal veriyi bozmamak için kopya üzerinde çalışıyoruz
islenmis_df =df.copy()

#yaş sütununundaki eskik değerleri yaşın medyanı ile dolduracağız
yas_medyani = islenmis_df['Age'].median()
islenmis_df['Age'] = islenmis_df['Age'].fillna(yas_medyani)

# Kabin sütunundaki eksik değerlere 'Unknown' atıyoruz
islenmis_df['Parch'] =islenmis_df['Parch'].fillna("Unkown")

# Kategorik değerleri Türkçeleştirme
islenmis_df['Sex'] = islenmis_df['Sex'].replace({'female':'Kadin','male':'Erkek'})

#Sınıf sütununu sayısal değil de bir kategori/etiket olarak algılaması için string yapıyoruz
islenmis_df['Pclass'] = islenmis_df['Pclass'].astype(str)

```

6-HARİTALAMA VE MERKEZİLEŞTİRME

```python
#vektörel merkezileştirme

#bilet ücretlerinin (Fare) genel aritmetik ortalamasını hesaplıyoruz
ortalama_ucret = islenmis_df['Fare'].mean()
#bilet ücretinden ortalama ücreti çıkarıyoruz
islenmis_df['Centered_Fare'] = islenmis_df['Fare'] - ortalama_ucret


#yaş ve cinsiyete göre yolcunun hangi grupta yer alacağını belirleyen fonksiyon
def yolcu_grubu(satir):
  if satir ['Age'] < 18:
    return 'Cocuk'
  elif satir ['Sex'] =='Kadin':
    return 'Yetiskin_Kadin'
  else:
    return 'Yetiskin_Erkek'

#yolcu_grubu fonkisyonunu DataFrame in tüm satırlarına uygular
#axis='columns satır satır çalışması ve sütun değerlerine erişim sağlar
islenmis_df['Yolcu_Grubu'] = islenmis_df.apply (yolcu_grubu, axis='columns')

print("\n Yeni Yolcu Grubu Dağılımı")
print(islenmis_df['Yolcu_Grubu'].value_counts())
#value_counts(), sütundaki benzersiz değerlerin kaç kez tekrarlandığını söyler
```

7- MODELLEME ÖNCESİ FEATURE VE TARGET AYRIMI

```python
#değişken belirleniyor
# modelin tahmin etmesini istediğimiz bağımlı değişkeni seçiyoruz
# 'Survived' sütunu yolcunun hayatta kalıp kalmadığını (0: Öldü, 1: Kurtuldu) temsil ediyoruz
y = islenmis_df['2urvived']

#girdiler seçiliyor
#modeli yolcunun durumunu tahmin etmek için kullanacağımız bağımsız değişkenleri seçiyoruz
#katkısı olmayan sütunları buraya dahil etmiyoruz
X = islenmis_df[['Pclass','Sex','sibsp','Parch','Fare','Embarked', 'Yolcu_Grubu']]


print("\n -Model Öcesi Kontrol-")

#1-Hedef değişkenin sınıf dağılımı
# normalize=True kullanarak hayatta kalanlar ile hayatını kaybedenlerin oranını yüzde olarak görüyoruz
#amacı veri setinde sınıf dengesizliği olup olmadığını anlamak
#örneğin verinin %99'u 0 olsaydı model sadece 0 tahmin ederek yanıltıcı yüksek başarı gösterebilirdi
print("y Dağılımı (sınıf dengesi kontrolü):")
print(y.value_counts(normalize=True))


#2-Girdi matrisinin x boyutu
# X.shape çıktısı satır sayısı, sütun sayısı formatında gelir
#amacı seçilen girdi sayısını doğrulamak ve satır kaybı olup olmadığını kontrol etmektir
print(f"\n2. Girdi Matrisi X Boyutu: {X.shape} ")


#3-eksik veri kontrolü
# X.isnull().sum() her sütundaki eksik değer sayısını verir,
#ikinci .sum() ise tüm tablodaki toplam eksik hücreyi sayar
#amacı birçok makine öğrenmesi algoritması NaN değerlerle çalışamaz
#çıktının 0 olması ve modelleme öncesi eksik veri kalmadığını garantiler
print("3. Kalan Eksik Veri Var Mı?: ", X.isnull().sum().sum())
```
